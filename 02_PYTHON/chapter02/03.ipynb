#### 대입 연산자(=)
x = 10
x = 10 + 20

연산의 우선 순위가 가장 낮다. 가장 마지막에 연산된다.

#### 산술 연산자
+, -
* : 곱하기
/: 나누기
//: 나눈 몫
%: 나눈 나머지
** : 2**3 -> 2 * 2 * 2
- 연산 우선순위: **, *, /, //, % > +, -
소괄호 () 로 묶어주면 우선순위 강제 적용


```python
num1 = 10
num2 = 3

num1 // num2, num1 % num2
```

```python
num1 = 3

num1 ** 3
```

#### 비교 연산자
>, <, >=, <=
== : 같다(동등성, 동일한 가치)
!= : 다르다

- 연산의 결과는 무조건 True, False
- 판별이 필요한 구문 (조건문, 반복문)

동등성 vs 동일성
- 동등성 : 동일한 가치를 비교(==)
- 동일성 : 동일한 대상인지 비교(is - 주소 비교)

```python
10 > 5
```

```python
num1 = 1000
num2 = 1000
id(num1), id(num2)
```

```python
num1 == num2 # 동등성 비교
```

```python
num1 is num2 # 동일성 비교(주소 비교)
```

```python
num3 = num1

num3 is num1
```

```python
num1 = 20

result1 = num1 > 10
result2 = num1 < 100

result1, result2
```

#### 논리 연산자 (and, or, not)

and
- 각 항이 모두 True -> True 하나라도 False 뜨면 False
or
- 각 항 중 하나라도 True -> True, 모두 False -> False

not
- True 조건 -> False 조건으로 반전, False 조건 -> True 조건으로 반전

비교 연산자 보다 우선순위가 낮다

```python
# and
num1 = 20

result1 = num1 > 10
result2 = num1 < 100

result1 and result2
```

```python
not (result1 and result2)
```

```python
fruit1 = "apple"
fruit2 = "pear"

result1 = fruit1 == "apple"
result2 = fruit2 == "orange"

result1, result2
```

```python
result1 or result2
```

```python
fruit1 == "apple" or fruit2 == "orange"
```

#### 부호 반전 연산자(-)

음수 -> 양수
양수 -> 음수

```python
num = 100

num2 = -num
num2
```

```python
num3 = -num2 # 음수 -> 양수
num3
```

```python
3. 단락 회로 평가(Short-Circuit Evaluation)
```

```python
result = True or 10**2
result
# 첫 항의 결과가 True -> or 연산의 결과가 이미 결정, 다음 연산 미진행
```

result = False or 10**2
result
# 첫 항의 결과가 False -> 결과가 아직 결정 X -> 다음 항 체크(연산)

```python
result = False and 10**2
result
```

```python
result = True and 10**2
result
```

```python
# or 연산을 이용한 기본값(없을 때 대체되는 값)
message = "" # 메세지가 값이 ""로 False 이므로 후열연산 시행

result = message or "기본 메세지"
result
```

```python
# and 연산을 이용한 간단한 조건 처리
height = 165

result = height >= 160 and "크다"
result
```
