- 판별식: 비교 연산자, 논리 연산자, 값

1. 단일 조건문: if

```
if 판별식:
    판별식이 참일 때 실행되는 코드
```

````python
num = 10
if num > 5:
    print("5보다 크다")
````

2. if-else 조건문

````python
num = 9
if num == 10:
    print("10 입니다.")
else: # if num == 10: 거짓일 때 실행되는 코드를 정의
    print("10이 아닙니다.")
````

3. 다중 조건 분기: if-elif-else

````python
age = int(input()) #값을 넣도록 유도하는 함수, 여기 기입된 값은 str이므로 숫자 쓸 땐 int 활용
print(age, type(age))
````

````python
age = int(input())
if age >= 20:
    print("성인")
elif age >= 17:
    print("고딩")
elif age >= 14:
    print("중딩") #앞선 조건이 참이면 앞에서만 실행되고 뒤는 계산 안함, 조건 부여에 주의
elif age >= 8:
    print("초딩")
else: #앞선 모든 조건문이 False일 경우 출력
    print("미취학아동")
````

4. 복잡한 조건 처리: 중첩 조건문과 논리 연산자 결합

````python
# 입력 받은 숫자가 10이상 100미만 인지 확인
num = int(input()) #효율적이진 않음, 가시성면에서도
if num >= 10:
    if num < 100:
        print(f"{num}은(는) 10이상 100미만 입니다.")
````

````python
# 논리 연산자 and 결합
num = int(input())

if num >= 10 and num <100:
    print(f"{num}은(는) 10이상 100미만 입니다.")
````
