2. 매개변수(Parameter)와 인수(Argument)

- 매개변수: 함수 호출시 입력 부분에 값을 담아주는 변수
- 인수: 호출시 실제로 사용된 값 (10)

3. 위치 인수(Positional Arguments)와 키워드 인수(Keyword Arguments)

- 위치 인수: 매개변수의 순서대로 값을 지정
- 키워드 인수: 매개변수 명과 함께 값을 지정, 순서는 중요하지 않음
    위치 인수와 키워드 인수 섞어 쓸 경우 위치 인수를 먼저 지정해줘야 오류 나지 x

````python
def add(num1, num2):
    
    return num1 + num2 # result를 정의하지 말고 그냥 return를 쓰면 간편
````

````python
add(10, 20) # 위치 인수
````

````python
add(num1=10, num2=20) # 키워드 인수 (직접 키워드를 통해 지정, 순서 바뀌어도 상관x)
````

````python
def add(num1, num2, name, mobile):
    return f"{num1 + num2}, {name=}, {mobile=}"
````

````python
add(10, 20, mobile="000-1000-1000", name="김영희")
````

5. 가변 인수(*args)와 키워드 가변 인수(**kwargs)

- *args: 가변적인 위치 인수, 매개변수 갯수 상관없이 지정 가능, 키워드 매개변수로는 사용 불가 ※args는 항상 kwargs 보다 앞서야함

````python
def add(num1, num2, num3, num4):
    return num1 + num2 + num3 + num4 # num 계속 추가 번거로움

add(10, 20, 30, 40)
````

````python
def add(*nums): #가변 인수로 한번에
    #print(nums, type(nums))
    total = 0
    for num in nums:
        total += num

    return total
````

````python
print(add(10, 20))
print(add(10, 20, 30))
print(add(10, 20, 30, 40))
````

````python
def print_user_info(**into):
    print(into)
````

````python
print_user_info(name="김철수", age=20)
````

````python
def print_user_info(name, age, **info):
    print("name:", name)
    print("age:", age)
    print(info)
    
````

````python
print_user_info("김철수", 20, mobile="010-1000-1000")
````

````python
def print_user_info(name, age, *args, **kwargs):
    print("name:", name)
    print("age:", age)
    print(args)
    print(kwargs)
````

````python
print_user_info("김철수", 20, mobile="010-1000-1000") #가변 인수들은 필수요소가 아니기 때문에 빈 상태로 출력이 된다.
````

````python
print_user_info("김철수", 20, 'A', 20, mobile="010-1000-1000") #'A', 20의 형태로 args에 출력되는 모습
````
