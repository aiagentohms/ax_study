- 변수 스코프(Scope) : 변수를 접근할 수 있는 범위

1. 전역 변수(Global)와 지역 변수(Local)

- 전역변수(Global): 함수 '외부'에 정의 변수, 코드 전역에서 접근 가능
- 지역변수(Local): 함수 '내부'에 정의된 변수, 함수가 실행될 때만 유효한 함수(콜 스택에서 로드)

````python
a = 10 # 전역 변수

def func():
    b = 20 # 지역 변수
    print(f"{a=},{b=}")
#print(a, b)
print(a)
func()
````

````python
c = 30

def outer(): #후행 범위 -> 선행 범위는 빼내올 수 있지만 반대는 안된다.
    a = 10
    print("c:", c)
    def inner():
        b = 20
        print(f"{a=},{b=}")

    inner()

outer()
````

2. 파이썬의 변수 검색 규칙: LEGB 스코프

접근 우선 순위: 1 -> 4
L(Local - 지역 변수) > E(Enclosing - 외부 함수 변수) > G(Global - 전역변수) > B(Buitin - 내장변수)

````python
a = 'Global'
def outer():
    a = 'Enclosing'
    print("outer:", a)

    def inner():
        a = 'Local'
        print("inner:", a)
    inner()
outer()
````

3. global 키워드

````python
a = 'Global'
def outer():
    a = 'Enclosing'
    print("outer:", a)

    def inner():
        print("inner:", a)
    inner()
outer()
````

````python
a = 'Global'
def outer():
    a = 'Enclosing'
    print("outer:", a)

    def inner():
        global a # a라는 변수는 전역변수를 찾겠다.

        print("inner:", a)

    inner()

outer()
````

````python
a = 'Global'
def outer():
    a = 'Enclosing'
    print("outer:", a)

    def inner():
        global a # a라는 변수는 전역변수를 찾겠다.
        a = not a
        print("inner:", a)

    inner()
    
outer()
````

4. 일급 합수(First-class Function)

- 함수와 변수를 동일하게 취급
- 함수의 매개변수로 함수를 사용할 수 있다! - 콜백 함수
- 함수의 반환값으로 함수를 사용할 수 있다! - 팩토리 함수, 클로저(Closure)

````python
def callback2():
    print("callback()")

def func(callback):
    print("func()")
    callback()

func(callback2)
````

````python
def outer(num1):
    def inner(num2):
        return num1 + num2

    return inner
````

````python
a = outer(10) #항상 10을 더해주는 새로운 함수
a(20) # 10 + 20
````

````python
b = outer(20) #항상 20을 더해주는 새로운 함수
b(20)
````

````python
def change_numbers(*args, callback): #(args는 튜플로 넘어옴, *은 함수 첫 작성 시 한번만, 작업하려면 리스트로 변경)
    numbers = list(args)
    
    for i in range(len(numbers)): #넘버의 총 개수만큼 반복
        numbers[i] = callback(numbers[i])

    return numbers
````

````python
def square(num):
    return num ** 2
````

````python
change_numbers(10, 20, 30, callback=square)
````

````python
def plus(num):
    return num + 10

change_numbers(10, 20, 30, callback=plus)
````

5. lambda 함수

- 간단하게 일회용으로 사용하는 경우 한줄의 식으로 간단하게 함수 정의
- 일회용으로 사용되므로 매개변수 이름도 최대한 간결하게(한자)
```
lambda 매개변수, 매개변수, ...: 처리 부분
```

````python
square = lambda num: num ** 2

square(10)
````

````python
def change_numbers(callback, *args):
    numbers = list(args)
    
    for i in range(len(numbers)):
        numbers[i] = callback(numbers[i])

    return numbers
````

````python
change_numbers(lambda x: x ** 3, 10, 20, 30) # 람다는 빌트인, 
````

7. lambda와 map(), fliter()를 활용한 데이터 일괄 가공

- map() : 변환작업
- fliter() : 특정 조건에서 데이터를 선별

````python
fruits = ['사과', '오렌지', '망고']
list(map(lambda x: f"**{x}**", fruits))
````

````python
numbers = [*range(1, 11)] #1 ~ 10 까지를 지시

list(filter(lambda x: x % 2 == 1, numbers))
````
