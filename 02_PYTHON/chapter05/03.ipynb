2. 클로저(Closure)

- 고차 함수(Higher Order Function)

```python
def outer():
    num1 = 10

    def inner():
        num2 = 20
        print(num1 + num2)
        
    return inner

inner = outer() #num1은 inner에 의해서 사용되므로 '콜스택'이 아닌 [힙 영역]에 생성
inner()
```

```python
def add(num1):

    def _add(num2):

        def __add(num3):
            return num1 + num2 + num3

        return __add
    
    return _add
    
```

3. 데코레이터(Decorator)

- 디자인 패턴 중 한가지

```python
def factorial1(num): #아래것 보다 이게 더 효율적. 이건 콜스택 한 번 쌓이고 끝나는데 아래건 10번
    result = 1

    for i in range(1, num + 1):
        result *= i

    return result
```

```python
import time #이 로직이 실행되고 끝나기 까지 시간 측정

start = time.time()
result = factorial1(10)
print(result)
end = time.time()
print("걸린 시간", end - start)
```

```python
#같은 함수 내에서 같은 함수를 반복해서 호출 하는 방법: 재귀 함수(Recursive)
def factorial2(num):
    if(num < 1):
        return 1 

    return num * factorial2(num - 1)
```

```python
import time

start = time.time() # 공통 기능
result = factorial2(10) #핵심 기능
print(result)
end = time.time() #공통 기능
print("걸린 시간", end - start)
```

```python
import time

def time_checker(callback, num):
    start = time.time() # 공통 기능

    result = callback(num) # 핵심 기능 - 팩토리얼 연산 함수(factorial1, factorial2)

    end = time.time() # 공통 기능
    print("걸린 시간:", end - start)

    return result
```

```python
time_checker(factorial2, 10)
```

```python
import time
# *args와 **kwargs 함께 넣어주면 범용적으로 인수를 넣어도 반응하게끔 만들 수 있다.
def time_check2(callback):
    def wrapper(*args, **kwargs):

        start = time.time() 

        result = callback(*args, **kwargs) # 핵심 기능

        end = time.time() #공통 기능
        print("걸린시간:", end - start)

        return result

    return wrapper
```

```python
@time_check2
def factorial2(num):
    if(num < 1):
        return 1 

    return num * factorial2(num - 1)
```

```python
factorial2(10)
```
