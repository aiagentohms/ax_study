```python
def get_total_and_average(*scores): # *은 맨 앞에만 찍으면 된다, 계속 찍으면 오류남
    if len(scores) == 0:
        return 0
    total = sum(scores)
    avg = total / len(scores)
    return total, avg
```

```python
result1 = get_total_and_average(90,85,95)
print(result1)
result2 = get_total_and_average(70,80,90,100,65)
print(result2)
```

```python
def peanut_sarang_gae(*sarang):
    if len(sarang) == 0:
        return 0
    total = sum(sarang)
    avg = total / len(sarang)
    return total, avg
```

```python
result1 = peanut_sarang_gae(90,85,100)
print(result1)
result2 = peanut_sarang_gae(70,80,90,100,65)
print(result2)
```
