```python
def message_store():
    storage = ""
    def add_message(message):
        nonlocal storage
        storage += message
        return storage
    return add_message
```

```python
my_storage = message_store()
my_storage("새로운 알림이 도착했습니다")
my_storage("출석 체크가 완료되었습니다")
```

```python
import time  # 시간 측정을 위한 표준 라이브러리 불러오기

# 실행 시간을 자동 연산하는 데코레이터 구현
def timer(original_func):
    # 어떤 인자든 유연하게 전달받을 수 있도록 가변 인자로 구성 
    def wrapper(*args, **kwargs):
        start_time = time.time()  # 함수 실행 전 시점 기록

        # 원본 함수를 가동하고 결과값을 안전하게 변수에 할당
        result = original_func(*args, **kwargs)

        end_time = time.time()    # 함수 실행 종료 시점 기록
        process_time = end_time - start_time  # 소요 시간 수치 계산

        print(f"[{original_func.__name__}] 함수 총 소요시간: {process_time:.6f}초")
        return result  # 원본 함수의 반환 결과 데이터를 최종 복귀시킴
    return wrapper

# 측정을 원하는 타겟 함수에 데코레이터 선언
@timer
def heavy_calculation(n):
    print(f"{n}까지의 정수 합산 반복 루프를 가동합니다.")
    total_sum = sum(range(1, n + 1))
    return total_sum

# 함수를 실행하여 성능 분석 결과 확인
calc_result = heavy_calculation(5000000)
print(f"최종 연산 결과값: {calc_result}")
```

```python
def deco_border(origin_func):
    def wrapper():
        print("=====")
        origin_func()
        print("=====")
    return wrapper
@deco_border
def show_welcome():
    print("환영합니다 스파르타님")
show_welcome()
```

```python
def auto_tracker(ori):
    def wrapper(*args, **kwargs):
        print(f"{ori.__name__}[함수 시작]")
        print(args)
        print(kwargs)
        result = ori(*args, **kwargs)
        print("연산이 종료되었습니다.")
        return result
    return wrapper

@auto_tracker
def add_numbers(a, b):
    return a + b

add_numbers(1, b=5)
```
