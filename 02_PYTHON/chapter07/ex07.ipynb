```python
menus = ["피자", "보쌈", "짜장면"]

try:
    index = int(input("조회할 인덱스 번호를 입력하세요: "))
    result = menus[index]
except ValueError:
    print("숫자만 입력할 수 있습니다.")
except IndexError:
    print("범위를 벗어난 위치입니다.")
else:
    print(f"조회가 성공했습니다. 결과값은 {result}입니다.")
finally:
    print("맛집 메뉴 조회를 종료합니다.")
```

```python
def shopping(amount):
    if amount <= 0:
        raise ValueError("주문 수량은 1보다 작을 수 없습니다.")
        print("수량 설정이 완료되었습니다.") 
    else:
        print("수량이 설정되었습니다.")
try:
    shopping(-1)
except ValueError as e:
    print(f"[시스템 경고] {e}")
```

```python
shopping(1)
```

```python
def set_user_age(age):
    if age < 0:
        # 비즈니스 규칙에 맞지 않는 음수 입력 시 의도적으로 예외를 발생시킴
        raise ValueError("나이는 음수가 될 수 없습니다.")
    print(f"사용자의 나이가 {age}세로 설정되었습니다.")

try:
    input_age = -5
    set_user_age(input_age)
except ValueError as e:
    # 발생한 예외를 식별하여 메시지 출력
    print(f"[시스템 경고] {e}")
```

```python
def calculate_share(total_price, num_people):
    share_per_person = total_price / num_people
    return share_per_person

def start_settlement():
    participants = 0
    result = calculate_share(50000, participants)
    print(result)

start_settlement()
```

```python
def calculate_revenue():
    total_revenue = 0
    daily_sales = [12000, 25000, 31000]
    for amount in daily_sales:
        total_revenue += amount
    print(total_revenue)

calculate_revenue()
```

```python
def sum_scores():
    final_score = 50
    bonus_scores = [10, 20, 30]
    for score in bonus_scores:
        final_score += score
    return final_score

result = sum_scores()
```
