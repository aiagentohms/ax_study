2. 클래스(Class)와 인스턴스(Instance)

```python
class Bus:

    # 속성 정의
    def __init__(self):
        self.count = 0 # 승객 수 - 초기화: 변수가 되었을 때 최초로 값을 대입
        self.passengers = []

    def ride(self, passenger): # 탑승
        self.count += 1 # 승객수 증가
        self.passengers.append(passenger)

        print("탑승")

    def off(self, passenger): # 하차
        if self.count > 1:
            self.count -= 1 # 승객수 차감
            self.passengers.remove(self.passenger)

        print("하차")

    def show_count(self):
        print("승객 수:", self.count)

    def print_self(self):
        print(id(self))

    def __str__(self):
        return f"승객 수: {self.count}, 승객: {self.passengers}"

class Passenger:
    # ※서로 다름 주의※
    # self.name, self.age -> 객체가 생성이 되면 만들어지는 인스턴스 변수
    # (..., name, age) -> 객체를 생성하기 위해서 매개변수로 정의된 지역 변수

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"이름: {self.name}, 나이: {self.age}"

    def __repr__(self):
        return f"Passenger(name='{self.name}', age = {self.age})"
```

```python
bus1 = Bus() #객체, 클래스 -> 객체(인스턴스 - 실체)

# 마침표(.) 연산자를 통해서 객체의 변수, 메서드에 접근 가능

print("탑승객 수:", bus1.count)
bus1.ride()
```

```python
bus2 = Bus() #bus1과 2가 가리키는 대상은 각각 다르다

bus1 is bus2 #같은 클래스로부터 생성된 객체는 서로 다른 객체!
```

```python
bus1 = Bus()
print("bus1 주소:", id(bus1))
bus1.print_self()
```

```python
bus = Bus()
p1 = Passenger()
p2 = Passenger() #p1, p2 같은 애들은 참조 변수, 객체가 아님

bus.ride(p1)
bus.ride(p2)

bus.show_count()
```

3. 초기화 함수 \_\_init\_\_ (생성자)

- 객체가 생성이 되면 가장 먼저 호출되는 메서드
- 객체가 생성된 직후에 해야할 처음 작업들(초기화 작업)
- 주로 하는 작업은 인스턴스 변수에 최초로 값을 할당(초기화)
- self 다음 매개변수는 객체를 생성할 때 외부에서 초기화 값을 지정할 수 있다.

4. self 키워드

- 클래스 내부에서 메서드의 첫 번째 매개변수로 정의
- 객체가 생성이 되면 객체의 자원을 접근하기 위한 참조 변수
- 클래스 내부에서만 사용 가능

5. 속성(Attribute)과 인스턴스 메서드(Method)

- 속성 : 객체마다 정의된 (인스턴스)변수
- 메서드: 객체가 가지고 있는 함수

6. \_\_str\_\_과 \_\_repr\_\_

- \_\_str\_\_ : print로 객체의 참조변수를 출력하면 자동으로 호출되는 메서드
    - 보통은 객체 내에 있는 인스턴스 변수의 내용을 출력하는 경우가 많다.

- \_\_repr\_\_ : print를 하면 자동으로 호출(\_\_str\_\_ 과 비슷), ※\_\_str\_\_ 의 우선 순위가 높다.
    - 로깅 목적(기록을 위한 목적, 디버깅용) : 클래스명(매개변수=값, ...) -> eval() 문자열을 실행했을 때 객체가 만들어지는 형태로 보통 정의
    - repr(객체 참조변수)

- eval(문자열) : 문자열을 파이썬 코드로 해석해서 실행 / 보안상 위험할 수 있으므로 사용에 유의

```python
bus = Bus()

p1 = Passenger("김철수", 20)
p2 = Passenger("김영희", 18)

bus.ride(p1)
bus.ride(p2)

print(bus)
```

```python
p1 = Passenger("김철수", 20)

print(p1) # __str__ 가 호출
print(repr(p1)) # __repr__ 가 호출
```
