```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
    def info(self):
        print(f"책 제목: {self.title}, 저자: {self.author}")

book_instance1 = Book("해리포터", "조앤K롤링")
book_instance2 = Book("반지의 제왕", "톨킨")


book_instance1.info()
book_instance2.info()
```

```python
class SmartLamp:
    def __init__(self, model):
        self.model = model
        self.is_on = False
    
    def turn_on(self):
        if self.is_on == False:
            self.is_on = True
            print(f"{self.model} 조명이 켜졌습니다.")

    def turn_off(self):
        if self.is_on == True:
            self.is_on = False
        print(f"{self.model} 조명이 꺼졌습니다.")

Lamp1 = SmartLamp("Lamp1")

Lamp1.turn_on()
Lamp1.turn_off()
```

```python
class Hero:
    def __init__(self, name):
        self.HP = 100
        self.name = name

    def take_damage(self, damage):
        self.HP -= damage
        print(self.HP)
        if self.HP <= 0:
            print(f"{self.name} 캐릭터가 쓰러졌습니다.")

Character1 = Hero("Character1")

Character1.take_damage(10)
Character1.take_damage(10)
Character1.take_damage(70)
```

```python
class Hero:
    def __init__(self, name):
        self.name = name
        self.HP = 100 #이건 매개변수 아닌 상수값

    def take_damage(self, damage):

        minus_damage = damage if self.HP >= damage else self.HP #삼항 조건
        self.HP -= minus_damage
        print("남은 체력:", self.HP)

        if self.HP <= 0:
            print(f"{self.name} 캐릭터가 쓰러졌습니다.")
```

```python
hero = Hero("피카츄")

hero.take_damage(90)
hero.take_damage(20)
```

```python
class Doorlock:
    def __init__(self, password):
        if not self.validate(password):
        self.__password = password

    def get_password(self):
        return self.__password

    def set_password(self, password):
        if not self.validate(self, password):
            return
        
        self.__password = str(password)

    def validate(self, password):

        result = len(str(password)) >= 4
        if not result:
            print("경고: 비밀번호는 4자리 이상이어야 합니다.")
        
        return result
```

```python
door = Doorlock(12)

door.__password = 'abcdef'
print(door.get_password())

door._Doorlock__password = 'abcdef'
print(door.get_password())
```
