1. 상속(Inheritance)

- 클래스 기반의 상속(상속에 대한 정의 -> 객체 생성(상속 관계 성립))

```
class 자식클래스(부모클래스):
    ... 
```

- super(self(참조)와 비슷함, ): 부모 클래스를 가리키는 참조 변수
    - 참고) self: 자식 쪽 객체(혹은 현재 객체)의 참조 변수
    - super() : 부모쪽 객체 생성

- 부모클래스 : 일반적인 정의들, 상위 개념
- 자식클래스 : 구체적인 정의들, 하위 개념

````python
class Animal:
    def __init__(self, kind): #이니트는 부모 클래스에 정의된 매직 메서드임
        self.kind = kind # 동물 종류

    def move(self):
        print("동물이 움직인다.")

class Dog(Animal):
    def __init__(self):
        super().__init__("강아지") # 부모 쪽 인스턴스 변수인 kind에 초기화

    def move(self):
        super().move() # 부모쪽 객체의 move()를 호출한다
        print(self.kind + " 걸음으로 움직인다.")

class Cat(Animal):
    def __init__(self):
        super().__init__("고양이")

    def move(self):
        print(self.kind + " 걸음으로 움직인다.")
    
````

````python
dog = Dog()
dog.move()

cat = Cat()
cat.move()
````

````python
dog = Dog()
dog.move()

cat = Cat()
cat.move()
````

````python
print(dog.kind, cat.kind)
````

2. 메서드 오버라이딩(Overriding)

- 메서드 재정의
- 부모쪽 클래스와 동일한 이름으로 정의하여 기능을 수정
- 자식쪽 클래스에 정의된 동일한 이름의 함수가 호출 우선순위를 가진다.
- 자식 쪽에서 부모쪽 동일한 이름의 메서드를 호출해야하는 경우는 self로 접근 x
    - super().부모쪽 메서드(..)로 접근

````python
class Human:
    def move(self):
        print("두 발로 직립보행한다.")
````

````python
human = Human()
human.move()
````

#### 다중 상속 - 여러 부모클래스를 상속 관계로 정의 가능

- 다중 상속 시 같은 메서드가 있을 때 호출 우선 순위(MRO)
- 자식 클래스의 메서드 -> 상속 관계에서 왼쪽에 있는 클래스 -> 

````python
class ParentA: #(항상 모든 클래스의 어버이는 object()이며 생략되어있음)
    def a(self):
        print("ParentA - a()")

    def b(self):
        print("ParentB - b()")

class ParentB:
    def a(self):
        print("ParentB - a()")

    def c(self):
        print("ParentB - c()")

# class Child(ParentA, ParentB):
#     pass

class Child(ParentB,ParentA):
    pass
````

````python
c = Child()
c.b()
c.c()
c.a()
````

````python
Child.__mro__ #Method Resolution Order - 메서드 우선순위
````

4. 정보 은닉과 캡슐화(Encapsulation)

````python
class Schedule:
    def __init__(self, year, month, day):
        self.year = year
        self.month = month
        self.day = day

    def __str__(self):
        return f"year= {self.year}, month={self.month}, day={self.day}"
````

````python
s1 = Schedule(2026,8,18)
print(s1)
````

````python
# instance 변수에 직접 대입이 가능하므로 값에 대한 통제는 불가!
# 통제 불가능하면 -> 직접 접근을 막아야 한다! 
s1.year = 2026
s1.month = 2
s1.day = 31

print(s1)
````

5. 파이썬 프라이빗(Private) 변수와 이름 변형 규칙(Name Mangling)

- \_\_변수명: 클래스 내부에서만 접근 가능한 인스턴스 변수가 된다.

````python
class Schedule:
    def __init__(self, year, month, day):
        self.__year = year
        self.__month = month
        self.__day = day

    def __str__(self):
        return f"year={self.__year}, month={self.__month}, day={self.__day}"
````

````python
s1 = Schedule(2026, 8, 18)

print(s1)
````

6. Getter와 Setter 메서드

````python
class Schedule:
    def __init__(self, year, month, day):
        self.__year = year
        self.__month = month
        self.__day = day
        
    def get_year(self):
        return self.__year

    def get_month(self):
        return self.__month

    def get_month(self):
        return self.__day

    def set_year(self, year):
        self.__year = year

    def set_month(self, month):
        self.__month = month

    def set_day(self, day):
        if self.__month == 2 and day > 28:    #값에 대한 [통제]가 가능
            day = 28

        self.__day = day

    def __str__(self):
        return f"year={self.__year}, month={self.__month}, day={self.__day}"
````

````python
s1 = Schedule(2026, 8, 18)

s1.set_month(2)
s1.set_day(31)

print(s1)
````
