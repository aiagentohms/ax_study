1. 클래스 변수(정적 변수), 인스턴스 변수

```python
class User:
    pass

user = User()
user.name = "김철수"
user.age = 20

print(user.name, user.age)
```

```python
class User:
    count = 0
```

```python
user1 = User()
user2 = User()

print(user1.count, user2.count)
```

```python
user1.count = 10

print(user1.count, user2.count)
```

```python
User.count = 100 # <- 이게 정적 변수이고 클래스 변수라고도 한다. 클래스변수는 모든 객체가 경유한다.
#인스턴스 변수와 착각할 수 있는 것을 클래스변수로 정의하고 접근한다.
print(User.count)
```

2. @classmethod, @staticmethod

- 객체를 생성하지 않아도 접근할 수 있는 메서드를 정의할 때

```python
class Calculator:
    count = 0

    # clz -> Calculator 클래스 객체 참조 변수
    # clz -> 클래스 변수, 다른 클래스 메서드를 접근하기 위한 목적
    @classmethod
    def add(clz, num1, num2):
        clz.count += 1
        print("호출 횟수:", clz.count)

        return num1 + num2

    @staticmethod
    def minus(num1, num2):
    
        return f"점수: {num1 - num2}"
```

```python
Calculator.add(10, 20)
```

```python
Calculator.minus(10,20)
```
