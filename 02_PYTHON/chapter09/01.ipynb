2. all()과 any()

- all(iterable) : 모든 요소가 참(True)로 인식될 때 -> True, 예외) 빈 리스트([]) 는 True로 받는다.
- any(iterable) : 요소 중 하나라도 True 인식이 되면 -> True

```python
all([True, 1, " "])
```

```python
all([False, False])
```

```python
all([]) # 거짓인 요소가 하나도 없으므로 참을 받음
```

```python
any([]) #참인 요소가 하나도 없으므로 거짓을 받음
```

3. len(), sum(), round()

- len(): 문자열 -> 문자열의 길이, 리스트, 튜플 -> 요소의 갯수
- sum(iterable) : 합계(요소는 숫자로만 구성)

```python
num = 10.56
print(round(num,1))

import math

print(math.ceil(10.2))
print(math.floor(num))
```

4. mean()과 median()

- statistics 모듈에 포함된 표준 라이브러리
- mean(iterable) : 평균
- median(iterable) : 중앙값

```python
import statistics

nums = [10, 15, 20, 8, 7, 1000000]

print("평균", statistics.mean(nums), "중앙값:", statistics.median(nums))
```

5. map()과 filter()

- map() : 변환
- filter() : 판별 -> 정제
-> 필터보단 리스트 컴프리헨션으로 대체(간단함) for - in 문

```python
nums = [*range(1, 11)]

nums2 = [x**2 for x in nums]
print(nums2)

nums3 = [x for x in nums if x % 2 == 1]
print(nums3)
```

6. enumerate() #열거하다 라는 뜻

```python
fruits = ['사과', '오렌지', '멜론', '망고']

num = 1
for fruit in fruits:
    print(num, fruit)
    num += 1
```

```python
for item in enumerate(fruits): # (순서 번호, 요소)
    print(item) #튜플로 합쳐져 나와서 쓰기 불편
```

```python
for item in enumerate(fruits): # (순서 번호, 요소)
    idx = item[0]
    fruit = item[1] #튜플 언패킹
    print(idx, fruit)
```

```python
for i, fruit in enumerate(fruits): # (순서 번호, 요소)
    
    print(i, fruit)
```

```python
for i, fruit in enumerate(fruits, start=1): # (1부터 시작)
    
    print(i, fruit)
```

7. zip()

참고 - itertools 모듈 zip_longest

```python
people = ["김철수", "김영희", "김민수"]
fruits = ["사과", "오렌지", "멜론", "망고"]
```

```python
#리스트 컴프리헨션
items = []
for i, peopl in enumerate(people):
    items.append((peopl, fruits[i]))

items
```

```python
items = zip(people, fruits)

list(items)
```

8. sorted() #원본을 바꾸지 않고 복사본을 만들어 비교하고 정리한다.

- 리스트.sort(): 원본 데이터를 정렬
- sorted(Iterable): 원본 데이터를 유지하고 새로운 객체를 생성해 정렬 (불변객체)

```python
import random

nums = [*range(1, 11)]
random.shuffle(nums)

nums
```

```python
nums.sort() #원본데이타(nums)가 정렬돼있음

nums
```

```python
nums_sorted = sorted(nums) #원본데이타(nums)를 건드리지 않고 정렬한 새로운 리스트를 만듦

print("nums_sorted:", nums_sorted)
print("nums:", nums)
```
