```python
class MyResource:

    def __enter__(self):
        print("__enter__ 호출")

        return self
    
# with 구문이 종료되었을 때 또는 예외가 발생했을 때 호출
# 예외가 발생시 exc_type - 예외 클래스 종류
#             exc_val - 예외 메세지
#             exc_tb - TraceBack
    def __exit__(self, exc_type, exc_val, exc_tb):
        print("__exit__호출")
```

```python
with MyResource() as o: # __enter__ 호출
    print(o)

    # 종료시 __exit__ 호출
```

```python
# 파일에서 작성하는 위치(파일 포인터)가 가장 앞으로 이동해서 내용을 대체
with open("test.txt", "w", encoding="UTF-8") as fp:
    fp.write("Hello!!")
```

```python
# 파일 포인터를 이동하지 않고 누적해서 작성( mode: a{append})
with open("test.txt", "a", encoding="UTF-8") as fp:
    fp.write("안녕하세요!\n")
```

2. 텍스트 파일 읽기 함수(read, readline, readlines)

- read(): 한꺼번에 전체 텍스트를 읽어온다.
- readline(): 한 라인 읽어오기
- readlines(): 라인별로 읽어서 리스트로 반환

```python
with open("test.txt", "r", encoding="UTF-8") as fp:
    text = fp.read()
    print(text)
```

```python
with open("test.txt", "r", encoding="UTF-8") as fp:
    # text = fp.readline()
    # print(text)
    # text = fp.readline()
    # print(text)
    # text = fp.readline()
    # print(text)
    # text = fp.readline()
    # print(text)
    # text = fp.readline()
    # print(text)
    while(text := fp.readline()): # 한줄씩 읽어서 메모리에 적재하므로 메모리 소비가 적다
        print(text)
```

```python
with open("test.txt", "r", encoding="UTF-8") as fp:
    lines = fp.readlines() # 한꺼번에 메모리에 오릴기 때문에 메모리 소비가 많다.
    print(lines)
```

```python
with open("test.txt", "r", encoding="UTF-8") as fp:
    for line in fp:
        print(line)
```

```python
3. CSV 형식
```

CSV - Comma Seperated Value
    - writer(파일포인터)
        = writerow(iterable)
    - reader(파일포인터)
    

```python
import csv

with open("members.csv", "w", encoding="UTF-8") as fp:
    writer = csv.writer(fp)
    writer.writerow(["이름", "나이", "전화번호"])
    writer.writerow(["김철수", 20, "010-1000-1000"])
```

```python
with open("members.csv", "r", encoding="UTF-8") as fp:
    reader = csv.reader(fp)
    members = list(reader)

    print(members)
```

4. JSON 형식

- 파이썬 객체 -> 전송 가능한 JSON 형식으로 변환 - 직렬화(Serialize)
- JSON 문자열(전송 받은 문자열) -> 파이썬 객체 변환 - 역직렬화

```python
import json

# dict -> JSON 파일 (직렬화)
user = {
    "name": "김영희",
    "age": 20,
    "mobile" : "010-1000-1000"
}

with open("user.json", "w", encoding="UTF-8") as fp:
        json.dump(user, fp, indent=4, ensure_ascii=False)
```

```python
# JSON 파일 -> dict (역직렬화)
with open("user.json", "r", encoding="UTF-8") as fp:
    user2 = json.load(fp)

    print(user2, "type:", type(user2))
```
