```python
with open("diary.txt", "w", encoding="UTF-8") as fp:
    fp.write("오늘은 너무 피곤했다")
    fp.write("잠을 더 일찍 자야겠다")
    fp.write("아침에 운동을 해야겠다")

with open("diary.txt", "r", encoding="UTF-8") as fp:
    content = fp.read()
    print("--- read() 결과---")
    print(content)
```

```python
import csv

fruitinfo1 = ["과일이름", "원산지", "과일색"]
fruitinfo2 = ["바나나", "필리핀", "노랑색"]
with open("data.csv", 'w', encoding="utf-8", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(fruitinfo1)
    writer.writerow(fruitinfo2)

with open("data.csv", "r", encoding="utf-8") as file:
    reader = csv.reader(file)
    data_list = list(reader)

print(data_list)
```

```python
import json

champ = {
    "캐릭터 이름": "Chara",
    "레벨": 99,
    "관리자 권한": True
}

with open("champ.json", "w", encoding="utf-8") as file:
    json.dump(champ, file)

with open("champ.json", "r", encoding="utf-8") as file:
    restored_champ = json.load(file)

print(restored_champ)
type(restored_champ)
```

```python
import matplotlib.pyplot as plt

month_income = {
    "month": ["January", "February", "March"],
    "income": [10000, 20000, 30000] 
}

df = pd.DataFrame(month_income)
print(df)
plt.plot(df["month"], df["income"], marker="o", label="Visitors")
plt.xlabel("Timeline")
plt.ylabel("Money")
plt.legend()
plt.grid(True)
plt.show()
```
