- 주피터 노트북에서는 이벤트 루프가 가동 중 -> await 함수명()
- 파이썬 파일(py)를 실행할 떄는 이벤트 루프를 생성해야 정상 동작

````python
import asyncio

async def work():
    await asyncio.sleep(2)
    print("작업 완료")
````

````python
await work()
````
