```python
import pandas as pd
```

```python
df = pd.read_csv("dataset.csv")
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
artists_mode = df["artists"].mode()[0]
df["artists"] = df["artists"].fillna(artists_mode)
```

```python
track_name_mode = df["track_name"].mode()[0]
df["track_name"] = df["track_name"].fillna(track_name_mode)
```

```python
album_name_mode = df["album_name"].mode()[0]
df["album_name"] = df["album_name"].fillna(album_name_mode)
```

```python
df.isnull().sum()
```

```python
duplicate_count = df.duplicated().sum()
print(f"중복된 데이터 개수: {duplicate_count}개")
```

```python
df = df.drop_duplicates().reset_index(drop=True)
print(f"중복된 데이터 개수: {df.duplicated().sum()}개")
df.describe()
```

```python
df = df[df['duration_ms'] < 1000000]
df = df[df['duration_ms'] >= 0]
```
