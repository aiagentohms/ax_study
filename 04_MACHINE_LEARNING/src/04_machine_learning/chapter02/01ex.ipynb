```python
import pandas as pd
df = pd.read_csv('usa_housing.csv')
over_2000 = df[df['YearBuilt'] > 2000]
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 5))
plt.scatter(over_2000['SchoolRating'], over_2000['Price'], color='dodgerblue', alpha=0.5, edgecolor='w')
plt.title('SchoolRating vs Price')
plt.xlabel('SchoolRating')
plt.ylabel('Price ($)')
plt.grid(True, linestyle='--')
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error
import sklearn
sklearn.set_config(display="text")
X=over_2000[['SchoolRating']]
y=over_2000[['Price']]


model = LinearRegression()
model.fit(X, y)
```

```python
y_pred = model.predict(X)
```

```python
print(model.coef_, model.intercept_)
```

```python
mse = mean_squared_error(y, y_pred)
mae = mean_absolute_error(y, y_pred)

print(f"컴퓨터가 찾은 기울기(w): {model.coef_}")
print(f"컴퓨터가 찾은 기본값(b): {model.intercept_}")
print(f"오차 제곱 평균(MSE): {mse:.2f}")
print(f"오차 절대값 평균(MAE): {mae:.2f}")
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='dodgerblue', alpha=0.5, edgecolor='w', label='Actual Data')
plt.plot(X, y_pred, color='crimson', linewidth=3, label='Regression Line')
plt.title('Linear Regression Prediction')
plt.xlabel('SchoolRating')
plt.ylabel('Price ($)')
plt.legend()
plt.grid(True, linestyle='--')
plt.show()
```
