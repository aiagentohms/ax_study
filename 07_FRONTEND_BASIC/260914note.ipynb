웹문서

<태그 이름> .. 내용 .. </태그이름>

여는 태그              닫는 태그

스타일링: CSS (Cascading Style Sheet) (확장자: .css)

동적요소: JAVAscript (확장자: .js)

웹 표준(Web Standard)

→ 어떤 브라우저를 사용해도 동일 화면, 동일 기능

2. 기본 구성

<!DOCTYPE html> → HTML 버전 태그 (HTML5)

<html> ← 문서 시작

<head> … </head> ⇒ 문서의 정보(title, meta ...), 문서 자원에 대한 정의

<body> … </body> ⇒ 문서에서 출력 부분

</html> ← 문서 종료

3. 태그(문서와 관련성이 높다)

1) 제목 태그 (Heading)
<h1> ~ <h6>
<h1> 주제목
<h2> 하위제목

2) 본문 서식 태그
<p> ... </p> -> 문단을 나눠준다 (paragraph)
<b> ... </b> -> 두껍게 만들어준다 (bold) = <strong></strong>
<i> ... </i> -> 기울임체로 만들어준다 (italic)
<u> ... </u> -> 밑줄을 그어준다 (underline)
<del> ... </del> -> 취소선을 그어준다 (delete) = <strike></strike>
<br> 줄개행 (단일 태그)

4. 주석

1) 설명

2) 출력 배제

3) 사용법: <!-- 설명 / 출력을 배제한 코드 -->

5. 문단 태그(Block 속성) ★중요
    1 - 줄개행
    2 - p, div, h1 ~ h6, form ...
    3 - 공간을 가진다.(너비, 높이를 설정할 수 있다)

6. 글 태그(Inline 속성) ★중요
    1 - 줄개행 X
    2 - i, b, strong, u, a, span
    3 - 공간을 가지지 않는다

하이퍼 링크(페이지 이름)

1. <a>: 하이퍼링크 태그
    태그의 추가 정보: 속성(attribute)
    -> 태그의 기능을 수행하기 위한 정보
    - href: 이동할 경로 - 절대 경로(전체 url or 전체 파일 경로, 경로가 변경이 되면 출력X)
                      - 상대 경로(./: 현재 파일이 있는 폴더 경로 or ../: 현재 폴더의 바로 위 상위 폴더)
    - target: 출력할 창의 위치
              _self(기본값) => 현재 창
              _blank => 새로운 창
              _parent => 부모 창 <iframe> y"자식창" - src: 창에 출력할 경로(url)
                                                    name: 창의 이름을 지어줌
              _top => 가장 상위에 있는 창
