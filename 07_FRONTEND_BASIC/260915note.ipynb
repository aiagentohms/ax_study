1장 3강

시멘틱 태그(sementic): 의미를 가지고 있는 태그
예) <form>, <table>, <video> ... <nav> 태그 이름만 봐도 어떤 기능을 하는지 알 수 있는 태그
*비시멘틱 태그: <div>, <span> ...

필요한 이유

1) 검색 엔진 최적화
2) 웹 접근성
3) 코드의 가독성

레이아웃을 구성하는 태그

1) <header>: 상단 공통 영역
2) <footer>: 하단 공통 영역
3) <main>: 본문, 하나만 정의
4) <section>: 내용 구분
5) <article>: 섹션별 내용
6) <nav>: 주 메뉴
7) <aside>: 왼쪽, 오른쪽 영역

2장 1강

CSS란?
Cascading(연쇄, 상속) Style Sheet - 꾸미는 언어
웹 문서의 스타일링 담당

1. 스타일 적용 기본 구조
선택자(selector) {
    속성: 값;
    속성: 값;
    ...
}

스타일 속성: color, background, width, height

2. 적용 방법
1) 인라인 스타일: 태그의 style 속성에 직접 적용
#ex  <ul>
        <li>사과</li>
        <li>배</li>
        <li style="color: orange; background: black;">감</li>
        <li>귤</li>
    </ul>
2) 내부 스타일 시트: <style>...</style>태그 안에 스타일을 정의 *지양해야할 스타일
<body>
    <style>
        li {color: blue; background: gray;
        }
    </style>
    <ul>
        <li>사과</li>
        <li>배</li>
    </ul>
3) 외부 스타일 시트: CSS파일을 생성하고 연동(<link ...>)
    <head>
        <meta charset="UTF-8">
        <meta name="viewport" content="width=device-width, initial-scale=1.0">
        <title>Document</title>

        <link rel="stylesheet" type="text/css" href="css/style.css">
    </head>

4) 주석
    /* ... #/
    설명
    해석 배제

5) 선택자 (selector)
    1 - 전체 선택자(*)
    2 - 태그 선택자 (태그 이름)
    3 - 클래스 선택자 - 여러개 요소 선택 / (.클래스명) / 복수개 선택 가능, 여러개 이름 지정(클래스)
    4 - 아이디 선택자 (#아이디명) - 아이디명은 하나만 지정 가능, 유일한 이름으로 지정
        (1개만을 선택하는 속성)

6) 스타일 적용 우선 순위
    범위가 넓을 수록 우선순위 ↓, 범위가 좁을 수록 우선순위 ↑
    태그 선택자 < 클래스 선택자 < 아이디 선택자 < Style 속성
      \_공통 st, 세부 st_/

      참고) 우선순위 강제: !important (남용X)

7) 복합 관계 선택자
    1 - 하위 선택자(조상-자손)
    ex)
    <body>
    <style>
        ol span {
            background: orange;
        }
    </style>
    <ul>
        <li>메뉴1</li>
        <li>메뉴2</li>
        <li>
            <span>메뉴3</span>
        </li>
        <li>메뉴4
            <ol>
                <li>서브 메뉴1</li>
                <li>
                    <span>서브 메뉴2</span>
                </li>
                <li>서브 메뉴3</li>
            </ol>
        </li>
        <li>메뉴5</li>
    2 - 자식 선택자(부모(바로 위 요소) > 자식(바로 아래 요소))

    3 - 일반 형제 선택자
        형-기준 ~ 동생-기준 아래에 정의된 요소들
    
    4 - 인접 형제 선택자
        형-기준 ~ 동생-기준 바로 아래 요소 1개
