1장 3강

시멘틱 태그(sementic): 의미를 가지고 있는 태그
예) <form>, <table>, <video> ... <nav> 태그 이름만 봐도 어떤 기능을 하는지 알 수 있는 태그
*비시멘틱 태그: <div>, <span> ...

필요한 이유

1) 검색 엔진 최적화
2) 웹 접근성
3) 코드의 가독성

레이아웃을 구성하는 태그

1) <header>: 상단 공통 영역
2) <footer>: 하단 공통 영역
3) <main>: 본문, 하나만 정의
4) <section>: 내용 구분
5) <article>: 섹션별 내용
6) <nav>: 주 메뉴
7) <aside>: 왼쪽, 오른쪽 영역

2장 1강

CSS란?
Cascading(연쇄, 상속) Style Sheet - 꾸미는 언어
웹 문서의 스타일링 담당

1. 스타일 적용 기본 구조
선택자(selector) {
    속성: 값;
    속성: 값;
    ...
}

스타일 속성: color, background, width, height

2. 적용 방법
1) 인라인 스타일: 태그의 style 속성에 직접 적용
예시)
    <ul>
        <li>사과</li>
        <li>배</li>
        <li style="color: orange; background: black;">감</li>
        <li>귤</li>
    </ul>
2) 내부 스타일 시트: <style>...</style>태그 안에 스타일을 정의 *지양해야할 스타일
예시)
<body>
    <style>
        li {color: blue; background: gray;
        }
    </style>
    <ul>
        <li>사과</li>
        <li>배</li>
    </ul>
3) 외부 스타일 시트: CSS파일을 생성하고 연동(<link ...>)
    예시)
    <head>
        <meta charset="UTF-8">
        <meta name="viewport" content="width=device-width, initial-scale=1.0">
        <title>Document</title>

        <link rel="stylesheet" type="text/css" href="css/style.css">
    </head>

4) 주석
    /* ... #/
