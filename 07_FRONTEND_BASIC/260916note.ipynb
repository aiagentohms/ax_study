1) 선택자 (selector)
    1 - 전체 선택자(*)
    2 - 태그 선택자 (태그 이름)
    3 - 클래스 선택자 - 여러개 요소 선택 / (.클래스명) / 복수개 선택 가능, 여러개 이름 지정(클래스)
    4 - 아이디 선택자 (#아이디명) - 아이디명은 하나만 지정 가능, 유일한 이름으로 지정
        (1개만을 선택하는 속성)

2) 스타일 적용 우선 순위

    태그 선택자 < 클래스 선택자 < 아이디 선택자 < Style 속성(자주 안씀)
      \_공통 st, 세부 st_/

      참고) 우선순위 강제: !important (남용X)

    예시)

        }

    6 - 속성 선택자 [속성명] : 지정된 속성이 있으면 선택
        1] 다른 선택자와 조합을 통해 선택 / 선택자[속성명]
        2] 속성명="값": 속성의 값이 일치하는 요소 선택
            속성명^="값": 값으로 시작하는 요소 선택
            속성명$="값": 값으로 끝나는 요소 선택 
            속성명*="값": 값이 포함된 요소 선택
    예시( 
        a[href^='tel'] {

3. 가상 클래스 선택자 [상태, 조건, 위치에 따라]
    1 - 가상 선택자
        (자식 요소는 타입이 달라도 선택이 된다.)
        : first-child  첫 번째 자식 요소
        : last-child   마지막 자식 요소
        : nth-child(n) 숫자 위치에 잇는 요소 선택, 1부터 시작
        :      " (수식) 수식에 해당하는 요소를 선택 ex(2n+1 => 홀수, 3n+1 => 세 번마다 한 번)
        (타입은 같은 타입들만 선택이 된다.)
        : first-of-type
        : last-of-type
        : nth-of-type
        : only-of-type -> 자식 요소가 1개만 있는 경우 선택
    2 - 조건 선택자
        ~ :is(선택자1, 선택자2, ...) 자식 선택자
            상위 요소 그룹 -> 하위 요소를 선택
        ~ :not(선택자) 
            지정된 선택자가 "아닌" 요소 선택
    예시(
         <style>
        li:not(.menu5) {
            background: orange;
        }
        :is(ul, ol) a {
            color: green;
        }
        </style>
        <ul>
            <li>메뉴1</li>
            <li>메뉴2</li>
            <li>메뉴3</li>
            <li>메뉴4</li>

            <li>메뉴6</li>
            <li>메뉴7</li>
            <li>메뉴8</li>
            <li>메뉴9</li>
            <li>메뉴10</li>
            <li><a>메뉴11</a></li>
        </ul>
        <ol>
            <li>메뉴1</li>
            <li><a>메뉴2</a></li>
        </ol>예시)

    3 - 상태 선택자
        : hover - 마우스가 요소에 올라간 상태
        : focus - 특정요소가 선택, 커서가 있는 상태
        : checked - 체크된 상태
        : read-only - 읽기 전용 상태
        : disabled - 비활성 상태
        : enabled - 활성 상태

    4 - 요소 추가 선택자
        선택자:: ___ (반드시 추가할 내용이 잇어야 노출 "content 필수"), inline 속성

        ::after  - 자식의 마지막 요소로 추가
        ::placeholder: 가이드라인 텍스트

```python
# 외부 여백과 내부 여백(margin, padding)

외부 여백(margin)
    - block 속성 태그: 제한 X
    - inline 속성 태그: 위 아래 여백은 적용 불가, 좌우 여백은 가능

    margin: 시계방향으로 공간을 채운다. 상 -> 우 -> 하 -> 좌; 순서
            상 좌우 하; 식으로도 가능, 좌우가 같다면 3개로 채운다
            상하 좌우; 이런 식으로도 가능, 위아래와 좌우를 채운다
            값; -> 모든 방향으로 동일하게 채운다

        top : 예시 - margin-top: 50px; -> 위로 50px 외부 여백, padding-top: 50px; -> 내부 여백
        bottom
        left
        right
        margin: 100px auto -> 위 아래 100px 좌우는 균등하게 여백 배분 (block 일때만 사용 가능)
        예시
            <style>
        .box {
            width: 150px;
            height: 150px;
            background: orange;
            /* margin: 10px 20px 30px 40px; */
            /* margin: 10px 20px 30px; */
            /* margin: 10px 20px; */
            /* margin: 100px; */
            margin: 100px auto;
        }
            </style>
    <div class="box"></div>
    <div class="box"></div>

Display
    1 - block: 이 아닌 속성을 block으로 바꿔준다.
    2 - inline: 이 아닌 속성을 inline으로 바꿔준다.
예시(
    <style>
        .box {
            width: 150px;
            height: 150px;
            background: orange;
            margin: 100px;
        }
        .box:nth-of-type(2) {
            background: blue;
        }
        .box:nth-of-type(3) {
            background: green;
        }

        .box:nth-of-type(4) {
            background: red;
        }

        div.box {
            display: inline;
        }

        span.box {
            display: block;
        }
    </style>
    <div class="box">A</div>
    <div class="box">B</div>

    <span class="box">C</span>
    <span class="box">D</span>
    예시)

    3. inline-block 줄개행 x - 공간 o

    4. none -> 감추기
```

Flexbox (1차원 정렬)
설정 - 부모요소 : display: flex;
적용 - 자식요소
정렬 방향
flex-direction
- row(기본값): 좌에서 우로
- row-reverse: 우에서 좌로
- column: 위에서 아래로
- column-reverse: 아래에서 위로

정렬 속성 제어 # 엄청 많이 쓴다.
justify-content
    - flex-start: 왼쪽 정렬(기본값)
    - flex-end: 오른쪽 정렬
    - center: 가운데 정렬
    - space-between: 좌, 우측의 요소 1개를 끝에 배치하고 나머지 요소들에 균등한 공백을 부여
    - space-around:
    - space-evenly
