#font-family

폰트명, 대체폰트1, 2, ...
폰트명에 공백이 있으면 따옴표로 감싼다.

예시: font-family: 'Malgun Gothic', 'Apple Gothic', sans-serif

시스템 폰트: os 내에 설치
웹 폰트: os에 설치 상관 없이 동일하게 공유 가능한 폰트
Chapter02, 16, 17.html

overflow
    visible -> 넘치는 컨텐츠가 있어도 보여줌
    hidden -> " 감춤
    scroll -> " 가 있으면 스크롤바 생성(상시 노출)
    auto -> " 있으면 스크롤바를 노출, " 없으면 미노출
    
border-width: 두께
border-style: none (기본값) solid (직선) dashed (파선) dotted (점선) double (이중실선)
border-color: 색상

단축 표현법
border: 두께 스타일 색상

background 배경
1. background-color: 배경색
2. background-image: url("이미지 경로")
3. background-repeat
    repeat: (기본값) x, y축 모두 반복
    repeat-x: x축 반복
    repeat-y:
    no-repeat:
4. background-position: 배경이미지 위치 설정
    행 x px 열 y px;
    left, right, center
    top, bottom, center
5. background-size: 가로 사이즈
6. background-attachment
    fixed -> 스크롤 해도 배경이미지는 고정
    scroll -> 스크롤 시 배경 이미지도 함께 스크롤
7. 단축 표기법
    background: 색상 이미지 <- 이 중 적어도 하나는 정의해야 함 / 반복 고정 위치 사이즈 (나열식으로 입력 가능) <- 선택>
