뷰포트(Viewport): 보이는 화면
width: 
    device-width - 장비의 너비 기준
    device-height - 장비의 높이 기준

user-scalable: yes - 확대, 축소 가능 no - 불가능
minimum-scale: 축소 비율
maximum-scale: 확대 비율
initial-scale: 처음 화면 비율

권장 사이즈: 해당 숫자들이 브레이크 포인트가 되며 'css적용이 달라진다'.
모바일 (~ 767px)
태블릿 (768 ~ 1023px)
데스크탑(1024px ~ )

미디어쿼리

@media 미디어 타입 and (미디어 특성) {
    "css 정의"
}

미디어 특성
    width / 
    height / 
    min-width: 크기 이상 / 
    max-width: 크기 이하
    orientation: 기기의 방향
        ㄴportrait: 세로
        ㄴlandscape: 가로

screen: 컴퓨터 모니터, 태블릿, 스마트폰
print: 프린트에만 해당하는 css
speech: "시각 장애인용 스크린 리더 css"
all: 기본값
