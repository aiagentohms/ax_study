### 2. 초기 자바스크립트의 변수 선언: var의 치명적인 결함

```
var a = 100;

-> 번역 과정 -> window 객체의 하위 속성으로 추가(웹브라우저 기준) / global 객체의 하위 속성으로 추가(Node.js)
```

1) "호이스팅 문제" : 정의한 변수, 함수가 최상위 객체에 할당된다.
    -> 함수 정의 위치에 상관없이 호출이 가능하다.

2) 변수의 유효범위(scope) -> '함수'

3) 선언자(var)을 생략하고 변수를 지역 변수로 정의 -> 전역 변수가 된다.


````javascript
var i = 10000; // 얘는 window.i / global.i = 100 

for (var i = 0; i < 10; i++) { // window.i, global.i = 0 ~ 10
    //반복 로직
    console.log("for문에 정의된 i:", i);
}

console.log(i); //window.i / global.i = 10
````

````javascript
console.log(a);

var a = 100;
````

### 4. 데이터 타입과 typeof를 활용한 타입 검증

#### 4.1 자바스크립트의 주요 데이터 타입
- Number (숫자형) : 정수와 실수를 따로 구분하지 않고 모든 숫자 형식
- String (문자열) : '', ""로 감싼 텍스트
- Boolean (불리언) : true, false
- null : 값이 없는, 혹은 비어 있는 상태임을 표시하는 값
- undefined : 정의되지 않음, 변수를 선언 후 값을 대입하지 않으면 기본값

````javascript
const num1 = 100;

console.log(typeof num1); // number

const str1 = 'ABC' ; //string
console.log(typeof str1);

const bool1 = true; // true, false
console.log(typeof bool1);
````

````javascript
let num2;

console.log(num2, typeof num2);
````

#### 4.3 문자열의 확장: 문자열 결합과 템플릿 리터럴 (ES6+)
