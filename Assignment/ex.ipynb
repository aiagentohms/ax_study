```python
# 가상의 회원 정보 데이터 집합 (리스트 내 딕셔너리 구조)
user_database = [
    {"name": "민수", "level": "VIP", "active": True, "point": 1500},
    {"name": "영희", "level": "Gold", "active": False, "point": 800},
    {"name": "철수", "level": "VIP", "active": True, "point": 2300},
    {"name": "지영", "level": "Silver", "active": True, "point": 400},
    {"name": "성민", "level": "Gold", "active": True, "point": 1200}
]

print("--- 마케팅 대상 활성 VIP 및 Gold 회원 식별 결과 ---")

for user in user_database:
    # 1단계 필터링: 계정의 활성화(active) 상태를 먼저 식별
    if user["active"]:
        # 2단계 필터링: 등급(level) 조건에 따른 다중 분기 제어
        if user["level"] == "VIP":
            print(f"[우수 등급] {user['name']} 회원님에게 VIP 전용 쿠폰이 발송되었습니다.")
        elif user["level"] == "Gold":
            print(f"[일반 등급] {user['name']} 회원님에게 Gold 기본 쿠폰이 발송되었습니다.")
```

```python
# 가상의 회원 정보 데이터 집합 (리스트 내 딕셔너리 구조)
user_database = [
    {"name": "민수", "level": "VIP", "active": True, "point": 1500},
    {"name": "영희", "level": "Gold", "active": False, "point": 800},
    {"name": "철수", "level": "VIP", "active": True, "point": 2300},
    {"name": "지영", "level": "Silver", "active": True, "point": 400},
    {"name": "성민", "level": "Gold", "active": True, "point": 1200}
]

# 조건별 데이터 누적 연산
total_vip_points = 0  # VIP 회원 포인트 합계 누적 변수
active_user_count = 0  # 활성 회원 수 카운트 변수
over_1000_count = 0 # 1,000점 이상 회원 카운트 변수

for user in user_database:
    if user["active"]:
        active_user_count += 1  # 활성 회원 개수 누적

        if user["level"] == "VIP":
            total_vip_points += user["point"]  # VIP 포인트 합산

    if user["point"] > 1000:
         over_1000_count += 1 # 1,000점 이상 회원 개수 누적

print(f"전체 활성 회원 수: {active_user_count}명")
print(f"활성 VIP 회원의 누적 포인트 합계: {total_vip_points}점")
print(f"1,000포인트 이상 회원 수: {over_1000_count}명")
```

```python
def get_average(*scores):
    if len(scores) == 0:
        return 0

    total = sum(scores)
    avg = total / len(scores)
    return avg

result1 = get_average(80, 90, 100)
result2 = get_average(75, 60, 88, 92, 100, 45)

print(f"3과목 평균 점수: {result1:.2f}")
print(f"6과목 평균 점수: {result2:.2f}")
```

```python
# 부가 기능을 주입할 데코레이터 함수 정의
def display_decorator(original_function):
    def wrapper():
        print("--- [데코레이터] 함수 실행을 시작합니다 ---")
        original_function()  # 인수로 받은 원본 함수를 식별하여 호출
        print("--- [데코레이터] 함수 실행이 완료되었습니다 ---")
    return wrapper  # 포장된 함수를 반환

# 원본 비즈니스 타겟 함수 정의
def core_logic():
    print("핵심 비즈니스 로직을 구현 중입니다.")

# 전통적인 방식의 데코레이터 연결 및 변수 매핑
decorated_logic = display_decorator(core_logic)

# 포장된 함수를 가동하여 전후 기능 확장 동작 확인
decorated_logic()
```
