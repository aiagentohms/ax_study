```python
class Book:
    def __init__(self, title, author, isbn):
        self.title = title
        self.author = author
        self.isbn = isbn
        self.is_borrowed = False

    def borrow(self):
        self.is_borrowed = True

    def return_book(self):
        self.is_borrowed = False
```

```python
book1 = Book(
    "파이썬 기초",
      "홍길동",
        "978-1234567890"
)

print(book1.title)
print(book1.author)
print(book1.isbn)
print(book1.is_borrowed)
```

```python
print(book1.is_borrowed)
book1.borrow()
print(book1.is_borrowed)
book1.return_book()
print(book1.is_borrowed)
```

```python
def add_book(books):
    title = input("도서명을 입력하세요: ")
    author = input("저자를 입력하세요: ")
    isbn = input("ISBN을 입력하세요: ")

    if isbn in books:
        print("이미 등록된 ISBN입니다.")
        return

    book = Book(title, author, isbn)
    books[isbn] = book

    print("도서가 등록되었습니다.")
```

```python
def search_book_by_isbn(books):
    isbn = input("검색할 ISBN을 입력하세요: ")

    if isbn in books:
        book = books[isbn]

        print("도서를 찾았습니다.")
        print(f"제목: {book.title}")
        print(f"저자: {book.author}")
        print(f"ISBN: {book.isbn}")
        print(f"대여 여부: {book.is_borrowed}")

    else:
        print("등록되지 않은 도서입니다.")
```

```python
def search_book_by_title(books):
    search_title = input("검색할 제목을 입력하세요: ")

    found = False

    for book in books.values():
        if book.title == search_title:
            print("도서를 찾았습니다.")
            print(f"제목: {book.title}")
            print(f"저자: {book.author}")
            print(f"ISBN: {book.isbn}")
            print(f"대여 여부: {book.is_borrowed}")

            found = True
    if found == False:
        print("도서를 찾을 수 없습니다.")
            
```

```python
def return_book(books):
    isbn = input("반납할 도서의 ISBN을 입력하세요: ")

    if isbn in books:
        book = books[isbn]

        if book.is_borrowed == False:
            print("현재 대여 중인 도서가 아닙니다.")
        else:
            book.return_book()
            print("도서가 반납되었습니다.")

    else:
        print("등록되지 않은 도서입니다.")
```
